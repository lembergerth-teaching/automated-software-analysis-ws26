# Lattice of Sets

The elements are all subsets of $s = \{1,2,3\}$, including the empty set and $s$ itself. This collection is called the power set of $s$. We order these elements by the superset relation: $x \sqsubseteq y$ if and only if $x \supseteq y$.

The following code generates the eight subsets. Each element is a Python set; the empty set is displayed as $\varnothing$.


```python
s = {1, 2, 3}


def get_elements(s):
    return [
        set(subset)
        for size in range(len(s) + 1)
        for subset in combinations(sorted(s), size)
    ]

elements = get_elements(s)
print(f"Elements: {elements}")
```

```python
def subsumption(x, y):
    return False


dot = Digraph()
relations = []
for x in elements:
    dot.node(str(x))
    for y in elements:
        if x == y:
            continue
        symbol = "⊑" if subsumption(x, y) else "⋢"
        relations.append(f"{x} {symbol} {y}")
        if subsumption(x, y):
            dot.edge(str(y), str(x))
for relation in relations:
    print(relation)
```
